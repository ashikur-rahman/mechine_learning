🧠Project: Image Classification with NN vs CNN (Fashion MNIST)

🎯 Objective:

Compare the performance of a basic Neural Network (NN) and a Convolutional Neural Network (CNN) on the Fashion MNIST dataset. Analyze which architecture performs better and why.

✅ Instructions:

📦 Prepare Your Dataset

    Use the Fashion MNIST dataset:

        Load using: tensorflow.keras.datasets.fashion_mnist

    Dataset details:

        60,000 training images, 10,000 test images

        28x28 grayscale images

        10 clothing item categories

    Preprocess the data:

        Normalize pixel values to [0, 1]

        Flatten images for NN input

        Reshape to 28×28×1 for CNN input

🧠 Build Two Models

1️⃣ Neural Network (NN)

    Build a simple fully connected network (e.g., using Dense layers only).

    Train on the training data.

    Evaluate on the test data.

2️⃣ Convolutional Neural Network (CNN)

    Build a CNN for the same classification task.

    Use Conv2D, MaxPooling, and Dense layers.

    Train and evaluate the model on the same dataset.

📈 Compare Performance

    For each model, display:

        Test Accuracy

        5 sample predictions (show image + predicted vs actual label)

    Use print() and visualizations to clearly show:

        NN test accuracy

        CNN test accuracy

        Sample output predictions for both

    (Optional) Plot training history (loss/accuracy) to show learning behavior

```python
# ==============================================================
# 🧠 Project: Image Classification with NN vs CNN (Fashion MNIST)
# ==============================================================

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import datasets, layers, models

# --------------------------------------------------------------
# 🎯 Step 1: Load and Prepare Dataset
# --------------------------------------------------------------
(fashion_train_images, fashion_train_labels), (fashion_test_images, fashion_test_labels) = datasets.fashion_mnist.load_data()

# Normalize pixel values to [0, 1]
fashion_train_images = fashion_train_images / 255.0
fashion_test_images = fashion_test_images / 255.0

# Class names for reference
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

print("Training data shape:", fashion_train_images.shape)
print("Test data shape:", fashion_test_images.shape)

# --------------------------------------------------------------
# 🧠 Step 2: Build a Basic Neural Network (NN)
# --------------------------------------------------------------
fashion_train_images_flat = fashion_train_images.reshape((60000, 28 * 28))
fashion_test_images_flat = fashion_test_images.reshape((10000, 28 * 28))

nn_model = models.Sequential([
    layers.Dense(128, activation='relu', input_shape=(784,)),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

nn_model.compile(optimizer='adam',
                 loss='sparse_categorical_crossentropy',
                 metrics=['accuracy'])

print("\n🔹 NN Model Summary:")
nn_model.summary()

# Train the NN
nn_history = nn_model.fit(fashion_train_images_flat, fashion_train_labels,
                          epochs=10, validation_split=0.2, verbose=2)

# Evaluate NN
nn_test_loss, nn_test_acc = nn_model.evaluate(fashion_test_images_flat, fashion_test_labels, verbose=2)
print(f"\n✅ NN Test Accuracy: {nn_test_acc*100:.2f}%")

# --------------------------------------------------------------
# 🧠 Step 3: Build a Convolutional Neural Network (CNN)
# --------------------------------------------------------------
fashion_train_images_cnn = fashion_train_images.reshape((60000, 28, 28, 1))
fashion_test_images_cnn = fashion_test_images.reshape((10000, 28, 28, 1))

cnn_model = models.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2,2)),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D((2,2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

cnn_model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

print("\n🔹 CNN Model Summary:")
cnn_model.summary()

# Train the CNN
cnn_history = cnn_model.fit(fashion_train_images_cnn, fashion_train_labels,
                            epochs=10, validation_split=0.2, verbose=2)

# Evaluate CNN
cnn_test_loss, cnn_test_acc = cnn_model.evaluate(fashion_test_images_cnn, fashion_test_labels, verbose=2)
print(f"\n✅ CNN Test Accuracy: {cnn_test_acc*100:.2f}%")

# --------------------------------------------------------------
# 📊 Step 4: Compare Performance
# --------------------------------------------------------------
print("\n📈 Performance Comparison:")
print(f"NN Accuracy  : {nn_test_acc*100:.2f}%")
print(f"CNN Accuracy : {cnn_test_acc*100:.2f}%")
print(f"🏆 CNN performed better by {(cnn_test_acc - nn_test_acc)*100:.2f}%")

# --------------------------------------------------------------
# 🖼️ Step 5: Display Sample Predictions
# --------------------------------------------------------------
def plot_sample_predictions(model, X, y, class_names, num_samples=5, is_cnn=False):
    plt.figure(figsize=(12, 4))
    for i in range(num_samples):
        plt.subplot(1, num_samples, i + 1)
        idx = np.random.randint(0, len(X))
        img = X[idx]
        img_input = img.reshape(1, 28, 28, 1) if is_cnn else img.reshape(1, 784)
        pred_label = np.argmax(model.predict(img_input, verbose=0))
        plt.imshow(img if not is_cnn else img.reshape(28, 28), cmap='gray')
        color = 'green' if pred_label == y[idx] else 'red'
        plt.title(f"P: {class_names[pred_label]}\nA: {class_names[y[idx]]}", color=color, fontsize=10)
        plt.axis('off')
    plt.tight_layout()
    plt.show()

print("\n🧠 NN Sample Predictions:")
plot_sample_predictions(nn_model, fashion_test_images, fashion_test_labels, class_names, is_cnn=False)

print("\n🧠 CNN Sample Predictions:")
plot_sample_predictions(cnn_model, fashion_test_images, fashion_test_labels, class_names, is_cnn=True)

# --------------------------------------------------------------
# 📉 Step 6: Plot Training History
# --------------------------------------------------------------
def plot_history(history, title):
    plt.figure(figsize=(12, 4))
    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'], label='Train Acc')
    plt.plot(history.history['val_accuracy'], label='Val Acc')
    plt.title(f'{title} Accuracy')
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'], label='Train Loss')
    plt.plot(history.history['val_loss'], label='Val Loss')
    plt.title(f'{title} Loss')
    plt.legend()

    plt.show()

plot_history(nn_history, "Neural Network")
plot_history(cnn_history, "Convolutional Neural Network")

```
